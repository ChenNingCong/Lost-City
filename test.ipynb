```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

# Sample list of data
data_list = rewards # 1000 random numbers from a standard normal distribution

# Create the figure and axes
plt.figure(figsize=(8, 5))

# Plot the smooth distribution using a KDE plot
# 'data' is the list/array of data
# 'fill=True' fills the area under the curve
# 'alpha' controls the transparency of the fill
sns.kdeplot(data=data_list, fill=True, alpha=0.5, linewidth=2, label='Smoothed Distribution (KDE)')

plt.title('Smooth Distribution of Data using Seaborn KDE')
plt.xlabel('Data Values')
plt.ylabel('Density')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
